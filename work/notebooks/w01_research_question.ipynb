## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*
**Decision:** which pages get reviewed for refresh, expansion, protection, pruning,
or monitoring in the next content cycle — and in what order.

**Who acts on it:** a content strategist or SEO lead allocating a small team's
limited writing and editing hours across a large page inventory.

**Cost of a wrong call:** asymmetric, and worth naming both sides.
- *False positive* (page flagged for refresh that did not need it): wasted writer
  and editor hours, plus the opportunity cost of the page that should have been
  worked on instead. Recoverable, but it erodes trust in the queue.
- *False negative* (a genuinely declining, valuable page that never surfaces):
  the decline continues unnoticed until someone catches it by hand, usually after
  real traffic has been lost.

Because review capacity is scarce, the top of the ranking matters far more than
overall accuracy — precision@K is the metric that reflects the real constraint.


```python
# Framing the evaluation constraint: only the top K pages get human review.
K = 20
print(f"Review capacity per cycle: top {K} pages -> precision@{K} is the metric that matters.")
```
